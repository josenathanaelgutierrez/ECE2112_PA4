```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('board2.xlsx')
```

A. VISAYAS COMMUNICATION DATAFRAME

```python

df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
```

```python

VisComm = df[(df['Hometown'] == 'Visayas') & (df['Track'] == 'Communication')][['Name', 'Gender', 'Math', 'Electronics', 'Average']]


display(VisComm)

print(f"Number of rows: {len(VisComm)}")
```

B. VISAYAS FEMALE DATAFRAME


```python
VisFemale = df[(df['Hometown'] == 'Visayas') & (df['Gender'] == 'Female')][['Name', 'Track', 'GEAS', 'Electronics', 'Average']]
display(VisFemale)
```

```python
display(VisFemale[VisFemale['Average'] >= 60])
```

C. CATEGORY-AVERAGE VISUALIZATION

```python
track_mean = df.groupby('Track')['Average'].mean().reset_index()
gender_mean = df.groupby('Gender')['Average'].mean().reset_index()
hometown_mean = df.groupby('Hometown')['Average'].mean().reset_index()
```

```python
display(track_mean)
```

```python
display(gender_mean)
```

```python
display(hometown_mean)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
y_max = 100

axes[0].bar(track_mean['Track'], track_mean['Average'], color='Blue')
axes[0].set_title('Mean Average by Track')
axes[0].set_xlabel('Track')
axes[0].set_ylabel('Mean Average')
axes[0].set_ylim(0, y_max)
axes[0].tick_params(axis='x', rotation=15)

axes[1].bar(gender_mean['Gender'], gender_mean['Average'], color='green')
axes[1].set_title('Mean Average by Gender')
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('Mean Average')
axes[1].set_ylim(0, y_max)

axes[2].bar(hometown_mean['Hometown'], hometown_mean['Average'], color='red')
axes[2].set_title('Mean Average by Hometown')
axes[2].set_xlabel('Hometown')
axes[2].set_ylabel('Mean Average')
axes[2].set_ylim(0, y_max)

plt.tight_layout()
plt.show()
```
